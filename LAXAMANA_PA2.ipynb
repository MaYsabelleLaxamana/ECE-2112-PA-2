### A. REPRODUCIBLE NORMALIZATION PROBLEM

This function generates a 5x5 array of reproducible random integers and normalizes the entire dataset by subtracting the array's mean and dividing by its standard deviation. 

```python
import numpy as np
#This creates the reproducible random 5x5 integer ndarray
np.random.seed(2112) 
X = np.random.randint(10, 101, size=(5, 5))

#This calculates the mean and standard deviation of X.
X_mean = np.mean (X)
X_std = np.std (X)

X_normalized = (X - X_mean)/X_std #This normalizes the array

print("X:\n", X)
print("Normalized X:\n", X_normalized)
print("Mean =", np.mean(X_normalized))
print("Standard Deviation =", np.std(X_normalized))

np.save("X_normalized.npy", X_normalized)
```

### B. CUBES DIVISIBLE BY 4 PROBLEM

This function creates a 10x10 array containing the cubes of the first 100 positive integers and uses Boolean masking to extract only the values that are evenly divisible by 4.

```python
#This creates the first 100 positive integers, cubes them, and reshapes them into a 10x10 array.
C = np.arange(1, 101)**3
C = C.reshape(10, 10)

div_by_4 = C[C%4 == 0] #This is where a Boolean condition is applied to obtain cubed values divisible by 4.

print("Shape of C:", C.shape)
print("Array div_by_4:\n", div_by_4)
print("Number of Selected Elements =", div_by_4.size)

np.save("div_by_4.npy", div_by_4)
```

### C. ABOVE-MEAN SQUARES PROBLEM

This function generates a 6x6 array containing the squares of the first 36 positive integers, calculates the overall mean of the array, and uses Boolean filtering to extract only the elements that are strictly greater than that mean.

```python
#This creates the squares of the first 36 positive integers and reshapes them into a 6x6 array.
S = np.arange(1, 37)**2
S = S.reshape(6, 6)

S_mean = np.mean(S) #This computes the mean of all elements of S.

above_mean = S[S > S_mean] #This uses Boolean filtering to select elements strictly greater than S_mean)

print("S:\n", S)
print("Mean of S:", S_mean)
print("Array above_mean:", above_mean)
print("Number of Selected Elements =", above_mean.size)

np.save("above_mean.npy", above_mean)
```
